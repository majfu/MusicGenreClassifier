```python
from pathlib import Path
from collections import defaultdict
import librosa
from tqdm import tqdm

WAV_FILES_FOLDER_PATH = '/content/drive/MyDrive/MGCProject/wav_audio_files'
```

**this example is for the FMA medium dataset**

------------------------------------

the most common sampling rate in the FMA dataset is 44100
___________________________

getting the length counts:

```python
def load_audio_file(wav_file_path):
    audio_signal, sampling_rate = librosa.load(wav_file_path, sr=44100)
    return audio_signal, sampling_rate

def get_length_counts(folder_path):
    folder_path = Path(folder_path)

    length_count_dict = defaultdict(int)
    length_ids_dict = defaultdict(list)

    wav_files = list(folder_path.rglob("*.wav"))

    for wav_file in tqdm(wav_files, desc="Processing WAV files"):
        track_id = int(wav_file.stem)
        audio_signal, sampling_rate = load_audio_file(wav_file)
        length_count_dict[len(audio_signal)] += 1
        length_ids_dict[len(audio_signal)].append(track_id)

    return length_count_dict, length_ids_dict

length_count_dict, length_ids_dict = get_length_counts(WAV_FILES_FOLDER_PATH)

print(length_count_dict)
```

length counts:

samples of length shorter than 1200000 will be removed

```python
threshold = 1200000

for length, ids in length_ids_dict.items():
    if length < threshold:
        print(f"Length {length}: {ids}")
```

for an audio sampled at 44100 Hz to be 30 s long, it needs to have:

30 * 44 100 = 1 323 000 samples

so the clips longer than that will be truncated and shorter than that will be zero padded
