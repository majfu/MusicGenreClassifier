getting the length counts:

```python
def load_audio_file(wav_file_path):
    audio_signal, sampling_rate = librosa.load(wav_file_path, sr=44100)
    return audio_signal, sampling_rate

def get_length_counts(folder_path):
    folder_path = Path(folder_path)

    length_count_dict = defaultdict(int)
    length_ids_dict = defaultdict(list)

    for wav_file in folder_path.rglob('*.wav'):
        track_id = int(wav_file.stem)
        audio_signal, sampling_rate = load_audio_file(wav_file)
        length_count_dict[len(audio_signal)] += 1
        length_ids_dict[len(audio_signal)].append(track_id)

    return length_count_dict, length_ids_dict

length_count_dict, length_ids_dict = get_length_counts(WAV_FILES_FOLDER_PATH)

print(length_count_dict)
```
